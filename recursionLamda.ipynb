```python
#factorial
def fact(n):
    if n ==0 or n==1:
        return 1
    return n*fact(n-1)
n=5
fact(n)
```

```python
def sum_of_natural(n):
    if n == 0:
        return 0
    return n*((n+1)/2) # or n+add(n-1)
n=int(input())
sum_of_natural(n)
```

```python
# fibonacci 
def fibonacci(n):
    if n<=1:
        return n
    else:
        return fibonacci(n-1)+fibonacci(n-2)
n = int(input())
for i in range(n):
    print(fibonacci(i), end=" ")
fibonacci(n)
```

```python
# reduce() retruns single value as a result
#if we want to work with reduce() we must import reduce from functools module
# from functools import reduce
# syntax :  reduce(lamda fun, iterable) 
from functools import reduce
l=[10,20,30,40,50]
x=reduce(lambda i,j:i+j,l)
print(x)
```

```python
d={'k':5,'x':10,'b':5,'t':1}
a=tuple(d.items())
r=sorted(a, key=lambda x:x[1])
print(r)
```
