```python
pin = "1804"
balance = 18004
attempts = 0
max_att = 3
Tx = []
# login functionality
while True:
    entered_pin = input("enter pin")
    if entered_pin == pin:
        print("PIN verification is successful..!")
        break
    else:
        attempts +=1
        print("Incoreect Pin, attempts lef :",(max_att - attempts) )
        if attempts > max_att:
            print("Your ATM card is locked")
            exit()
# core functionality: Menu
while True:
    print("🏧--- ATM Menu ---🏧")
    print("1. Check Balance")
    print("2. Deposit Money")
    print("3. Withdraw Money")
    print("4. No. ofTranctions (last 4 Tx)🔃")
    print("5. exit")

    choice = input("enter your choice 😊:")
    if choice == "1":
        print("your current balance is:", balance,"💵")
    elif choice == "2":
        deposit_amount = int(input("enter amount for deposit🔝"))
        if deposit_amount > 0:
            balance = balance + deposit_amount
            Tx.append(f"deposit is {deposit_amount}")
            if len(Tx)>5:
                Tx.pop()
                
            print("Deposit successfull, New Balance :", balance)
        else:
            print("invalid amount")
    elif choice == "3":
        withdrawn_amount = int(input("enter amount to withdrawn :🔜"))
        if withdrawn_amount > 0 and withdrawn_amount <= balance:
            balance = balance - withdrawn_amount
            Tx.append(f"withdrawn:{withdrawn_amount}")
            if len(Tx)>5:
                Tx.pop(0)
            print("withdrawn successfull, New balance : ", balance)
        else:
            print("insufficent balance or invalid input")
    elif choice == "4":
        print("current Transactions 🔃:",)
        if len(Tx)==0:
            print("no Tx are done")
        else:
            for t in Tx:
                print(t)
    elif choice =="5":
        print("BYE....😊 ")
        break
    else:
        print("invalid go back to home")
```
