```python
# ...> math
# used to perform basic mathamatical caluclations
# in math module we have functions like --> Sqrt, factorial, pow, pi, ceil, GCD, LCM

import math
print(math.sqrt(125))
print(math.sqrt(9))
print(int(math.sqrt(36)))
print(math.isqrt(9))

print(math.pi)

print(math.factorial(5))

print(math.pow(3,2))
print(int(math.pow(3,2)))

print(math.gcd(10,20))
print(math.lcm(7,45))

# ceil - if a number have decimal value i.e .0 value then it will return least integer
#       if a number have decial value >.1 then it will return greatest integer

# round - If a number have decimal value of <=.4 it will return least integer
#       if a number have >=.6 it will return gratest integer
#       if a number have exact .5 deciaml it will return nearest ~even integer~

print(math.ceil(3.0))
print(math.ceil(3.1))
print(math.ceil(3.5))

print(round(3.0))
print(round(3.4))
print(round(3.6))
print(round(2.5))
```

```python
# ...> RANDOM
# it is used to generate random values
# we use randint, choice functions in random modeule

import random
print(random.randint(1,100))

# how to generate 4 digit OTP

for i in range(6):
    print(random.randint(0,9), end="")
```

```python
#...> SYS
# used to know system related data

import sys
print(sys.path)
print(sys.version)
l=[1,3,4]
t=(1,3,4)
print(sys.getsizeof(l))
print(sys.getsizeof(t))
```

```python
# ...> OS
import os
print(os.name)
print(os.getcwd()) # get current working directory
print(os.listdir())
```

```python
# date time

import datetime
print(datetime.datetime.now())
print(datetime.datetime.now().year)
print(datetime.datetime.now().month)
print(datetime.datetime.now().day)
print(datetime.datetime.now().hour)
print(datetime.datetime.now().minute)
print(datetime.datetime.now().second)
print(datetime.datetime.now().microsecond)
print(datetime.datetime.today() - datetime.timedelta(days=1))
print(datetime.datetime.today() - datetime.timedelta(hours=1))
print(datetime.datetime.today() - datetime.timedelta(minutes=1))
print(datetime.datetime.today() - datetime.timedelta(seconds=1))
print(datetime.datetime.today() - datetime.timedelta(microseconds=1))
print(datetime.datetime.today() + datetime.timedelta(days=1))
print(datetime.datetime.now().strftime("%A")) # returns day
print(datetime.datetime.now().strftime("%B")) # returns month
print(datetime.datetime.now().strftime("%Y")) # returns year
print(datetime.datetime.now().strftime("%H")) # returns hour
print(datetime.datetime.now().strftime("%M")) # returns minute
print(datetime.datetime.now().strftime("%S")) # returns second
```

```python
#..> Statistics
# used to know MEAN, MODE, MEDIAN

import statistics
l=[2,6,9,4,5,2,8,0,5,2]
print(statistics.mean(l))
print(statistics.median(l))
print(statistics.mode(l))
print(statistics.stdev(l))
print(statistics.variance(l))
print(statistics.pstdev(l))
print(statistics.pvariance(l))

```
