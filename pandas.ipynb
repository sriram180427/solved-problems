*PANDAS*

Pandas are mainly used to handle large amount of data.
We can use pandas for data analytics, data cleaning
Pandas supports Tabular format results
Pandas are Faster than Lists
 Mainly 2 fns used :
        1. Series..> single comma representation
        2. DataFrame..> used to represent multiple coloumns and tables
If we want to work with pands we need to install Pandas using cmd ( pip install pandas)

then while working with Pandas we need to import padas (import pandas as pd)


```python
import pandas as pd
res = pd.Series([10,20,30,40,50,60,70], index = ["ram","kart","r","t","y","u","i"])
print(res)
```

```python
s = pd.DataFrame({"product": ["Mobile","Tab","TV","AC","Lap","Wash"],
                  "price": [19000,32000,382390,467380,4570,6758490],
                  "quant":[15,6,8,4,7,4]})
print(s)
```

```python
print(s.shape) # returns rows and colums
print(s.columns) # returns name of columns
print(s.head(3)) # by default returns top 5 rows or if value mentioned those top values returned
print(s.tail(2)) # returns rows from bottom
```

```python
s = pd.DataFrame({"product": ["Mobile","Tab","TV","AC",None,"Wash"],
                  "price": [19000,32000,382390,467380,4570,None],
                  "quant":[15,6,8,None,7,4]})
print(s)
print(s.fillna("Hello")) # fills null value
print(s.dropna()) # drops null value rows
print(s.info()) # gives table structure
print(s.describe()) #
```
