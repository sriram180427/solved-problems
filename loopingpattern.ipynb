```python
# find number of digits in a number using while
# reverse of a given number
#check a given  num is palindrome or not
# sum of digits in a num
```

```python
# 1. no of digits 
x = int(input())
count = 0
while x >0:
    x //=10
    count += 1
print(count)
```

```python
# 2. rev of num
x = int(input())
r=0
while x>0:
    num = x%10
    r = r*10 + num
    x = x//10
print(r)
```

```python
# 3. check given num is palindrome or not
x= int(input())
y = x 
r =0
while x >0:
    num = x % 10
    r = r * 10 + num
    x= x // 10
print("palindrome" if r == y else "not")
    
```

```python
# sum of digits
x = int(input())
sum = 0
while x >0:
    num = x % 10
    x = x // 10
    sum = sum+ num
print(sum)
```

```python
# ...> NESTED LOOPs  / Patterns
# triangle at right gaps
n = int(input())
for i in range(0,n+1):
    for j in range(i):
        print("*",end = " ")
    print()
# reverse triangle right gaps
n = int(input())
for i in range(n,0,-1):
    for j in range(i):
        print("*",end = " ")
    print()

# trianglee with left gaps
n = int(input())
for i in range(1,n+1):
    for j in range(n-i):
        print(" ",end=" ")
    for k in range(i):
        print("*", end=" ")
    print()

# reverse triangle with left gaps

n = int(input())
for i in range(n):
    for j in range(i):
        print(" ",end = " ")
    for k in range(n-i):
        print("x", end = " ")
    print()

    
```

```python
# butterfly pattern
n= int(input())
for i in range(1,n+1): # uppeer half
    for j in range (i):
        print("*", end = " ")
    for k in range(2*(n-i)):
        print(" ", end = " ")
    for j in range (i):
        print("*",end=" ")
    print()
for i in range(n, 0 ,-1): # lower half
    for j in range(i):
        print("*", end = " ")
    for k in range(2*(n-i)):
        print(" ", end= " ")

    for j in range(i):
        print("*", end = " ")
    print()
            
```
