```python
# Set 
#immutable Set elements but set is mutable 
# collection of unique elements and dont allow repetation / duplicates
#unordered data type
# empty set= set()
# indexing and Slicing is not allowed in Sets
s={2,7,4,"ram","netha",(1,9,6)}
print(s)
s.clear() # empty set
print(s)
k={1,4,6,7,4,2,1,8} # doesnot allow repetations
print(k)

      # BUILT IN METHODS 
#add(), update(), remove()/discard(), pop(), clear(), copy()

E = {10,60,40,30,55,66,60,10}
print(E)
E.add(459) # only one eleemt can be added no multiple  elements can be added using add()
print(E)
E.update({45,23,"sr",367}) # we can add multiple elements using this method i.e by using another set inside function
print(E)
E. remove(30) # removes particular elemnets
print(E)
E. pop()
print(E)
E.pop() # removes 1st elemment on set
print(E)
E.clear()
print(E)

# Built-in Functions 
# Len(), min(), max(), sum(), sorted()

# Set operations 
# .union(), .intersection(), .difference(), .symmetric_diffrence(), symmetric_diffrence_update()

s1= {1,2,3}
s2= {3,4,5}
s1.union(s2)
print(s1 | s2)
s1.intersection(s2)
print(s1 & s2)
s1.difference(s2)
s1.symmetric_difference(s2)

s1.symmetric_difference_update(s2)
print(s1)

# Frozen set 
#immutable version of set represented as Frozenset()
# it can be used as keys
s="pyhton"
k=frozenset(s)
print(k)
```
