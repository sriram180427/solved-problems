*MATPLOT LIB*

Matplotlib is used to visualization of Data
1. Figure -> The Whole Window
2. Axes -> x-axis, y-axis
3. Label -> Names of axis
4. Legend -> used in pichart
5. Title ->  entire Visualization Name
mainly we have 4 types of visualization
1. Line Ploit
2. Bar Chart
3. Pie Chart
4. Histogram Chart

```python
# 1. Line Plot
import matplotlib.pyplot as p
years = [2018,2019,2020,2021,2022,2023,2024]
admin = [600,200,400,900,1000,1100,100]
p.plot(years,admin)
p.title("Admissions")
p.xlabel("Years")
p.ylabel("Admin")
p.grid()
p.show()
```

```python
# Bar Chart
p.bar(years,admin)
p.title("Admissions")
p.xlabel("Years")
p.ylabel("Admin")
p.grid(False)
p.show()
```

```python
#Histogram
Ages = [10,19,17,15,16,15,11,15,19,14]
p.hist(Ages)
p.show()
```
