```python
# Static Method
# static method has no realtion with class and object
# the 1st parameter of static method is neither self nor cls
#it is a utility or helper method in class
# to make a method as static method we must write @staticmethod decorator above method

class Student:
    @staticmethod
    def Greet():
        print("Welcome to Ram")

s= Student()
s.Greet()
```

```python
# ...>Single Inheritance
# A Single child acuries properties from single parent class

class Animal:
    def Sound(self):
        return " Animals making Sound"
class Dog(Animal):
    def bark(self):
        return " Dog is barking"
d = Dog()
print(d.bark())
print(d.Sound())
```

ABSTRACTION

```python
from abc import ABC, abstractmethod
class Mobile(ABC):
    @abstractmethod
    def camera(self):
        pass
    @abstractmethod
    def display(self):
        pass
    @abstractmethod
    def memory_space(self):
        pass

    def details(self):  #contrete  method
        return " Hello Vivo"

class Vivo(Mobile):
    def __init__(self, model,quality):
        self.model = model
        self.quality = quality
    def camera(self):
        return (f"mu mobile camera quality is {self.quality}")
    def display(self):
        return (f"my mobile display quality is {self.quality}")
    def memory_space(self, Ram, Rom):
        return (f"my mobile Ram is {Ram}, rom is {Rom}")

v1 = Vivo("1150", 1080)
print(v1.camera())
print(v1.display())
print(v1.memory_space(4, 256))
print(v1.details())
```

Exceptional Handling

```python
# try, except, finally
# try is used to write risky code

print("Start")
try:
    a = int(input())
    b = int(input())
    print(a+b)
    print(a-b)
    print(a*b)
    print(a/b)
except ZeroDivisionError:
    print("give proper integer")
print("end")
```

```python
#name error
try:
    age = 25
    print(age)
except NameError:
    print("give proper name")
```

```python
#type error
try:
    l = [10,20,30,40]
    print(l[0])
    print(l[-1:])
    print(l[4])
except IndexError:
    print("invalid index")
```

```python
print("Start")
try:
    a = int(input())
    b = int(input())
    print(a+b)
    print(a-b)
    print(a*b)
    print(a/b)
except Exception as msg:
    print(msg)
    print(type(msg))
print("end")
```

```python
# Finally
# it will always execute even if get or dont get error
#closing purpose
print("Start")
try:
    a = int(input())
    b = int(input())
    print(a+b)
    print(a-b)
    print(a*b)
    print(a/b)
except Exception as e: # if error occurs
    print(e)
else: #if  no error occur in try blocck else will execute
    print("give proper integer:",(a+b))
finally:
    print("final block")
print("end")
```
