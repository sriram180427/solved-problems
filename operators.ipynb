```python
#Assignmet operators
a=10
a+=5
print(a)
a-=3
print(a)
a*=5
print(a)
a%=5
print(a)
a=10
a/=10
print(a)
a=15
a//=2
print(a)
```

```python
#comparison or Relational operators
print(10==7) # compares to values
print(10>7)
print(10<7)
print(10<=15)
print(10>=7)
print(10!=7)
```

```python
#Logical operators
# and, or , not
#used to compare 2 conditions and return true/false
print(10>7 and 10!=7)
print(10<7 or 10!=7)
print(10<7 and 10!=7)
print(not(10>7))
```

```python
# Identity operator
#used to compare two  value address and return true/false

#case 1. a=10  b=20
a=10
b=20
print( a == b) # compares only the values
print(id(a)) # prints address of a
print(id(b)) # prints address of b
print( a is b)  # compares wether the address of both values are same or not

# case 2, a=10  b=10
a=10
b=10
print(id(a)) # prints address of a
print(id(b)) # prints address of b
print( a is b)  # compares wether the address of both values are same or not

# case 3, using list  a=[1,2,3]  b=[1,2,3]
a=[1,2,3] 
b=[1,2,3]
print(id(a)) # prints address of a
print(id(b)) # prints address of b
print( a is b)  # compares wether the address of both values are same or not
```

```python
# Membership operator
# it is used to check wether the element exists or not in itterable
a=[10,20,30,40]
print(10 in a)
s="Sri Ram"
print(" "in s)
x={1,"s",10}
print(1 in x)
```

```python
#Bitwise operators 
#converts numbers into binary values and perform binary operations 
# & (AND)
# | (OR)
# ~ (NOT) -(x+1)
# ^ (XOR)
# >> (right shift)  x//2power(n)
# << (left shift)  x*2power(n)
a=5
b=1
print(a&b)
print(a|b)
print(a^b)
print(~b)
print(a<<b)
print(a>>b)

```
